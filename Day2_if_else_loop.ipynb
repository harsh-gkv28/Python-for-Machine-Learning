1. Operators in Python

* Arithmatics Operators
* Relational Operator
* Logical Operator
* Bitwise Operator

```python
# Arithmatics Operator
print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2)  # Integer duvision

print(5%2) # Modulas operator

print(5**2)  # Power off Operator
```

```python
# Relational Operator
print(4>5)

print(4>=5)

print(4<5)

print(4>=5)

print(4<=4)

print(4==4)

print(4!=4)
```

```python
# Logical Operator (And, OR,Not)

print(1 and 0)

print(1 or 0)

print(not 1)

```

```python
# Bitwise Operator - Special operator jo Binary Value par operate Karta ha

print(2 & 3)    # Bitwise And Operator

print ( 2 | 3)  # Bitwise Or Operator

print(2 ^ 3 )   # Bitwise XOR Operator

print(~ 3)      # Bitwise Not Operator

print(4 >> 2)  # Left Left Shift Operator

print (5 << 2 ) # Bitwise Right Shift Operator
```

```python
# Assigment Operator =

# a = 2 equal to is assigment operator

a = 2
# a = a + 2
a = a+2
print(a)
```

```python
# Membership Operator
print('D' in  'Delhi')


print('D' not in 'Delhi')

print(1 in [2,3,4,5,6,])
```

```python
# Menu Driven Calculator
fnum = int(input('enter the 1st number'))
snum = int(input('enter the 2nd number'))

op = input('enter the operation')

if op =='+':
  print(fnum + snum)
elif op =='-':
  print(fnum - snum)
elif op =='*':
  print(fnum * snum)
else:
  print(fnum / snum)
```

6. Loop Control Statement
* Break Statement
* Continue

```python
# Break Statement
for i in range(1,10):
  if i == 5:
    break
  print(i)
```

```python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))

for i in range(lower,upper+1):
 for j in range(2,1):
  if i%j== 0:
    break
 else:
    print(i)
```

```python
#continue Statement
for i in range(1,10):
  if i == 5:
    continue
  print(i)
```
